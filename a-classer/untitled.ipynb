```python
import matplotlib.pyplot as plt
import numpy as np

# Données de base
parametres = ['Taux d\'intérêt', 'Actions', 'Spread crédit',
              'Mortalité', 'Rachat']
scr_base = [12, 64.4, 2.9, 18.0, 15.0]  # en M€

# Scénarios de stress supplémentaires
scr_stress = [8.5, 80.0, 4.2, 24.0, 22.0]  # en M€

# Calcul de l'impact
impact = np.array(scr_stress) - np.array(scr_base)

# Graphique tornado
fig, ax = plt.subplots(figsize=(10, 6))
colors = ['red' if x > 0 else 'green' for x in impact]
ax.barh(parametres, impact, color=colors, alpha=0.7)
ax.axvline(0, color='black', linewidth=0.8)
ax.set_xlabel('Impact sur le SCR (M€)', fontsize=12)
ax.set_title('Analyse de sensibilité du SCR', fontsize=14, fontweight='bold')
ax.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```
